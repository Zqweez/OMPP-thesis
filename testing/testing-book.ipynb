```python
from modlamp.descriptors import GlobalDescriptor

seqs = ["FLPLFLPKIICVITKKC"]

gd = GlobalDescriptor(seqs)

# Calculate charge directly default pH 7 and amide = False
gd.calculate_charge() 
charge_direct = gd.descriptor.copy()

# Calculate charge with calculate_all default pH 7.4 and amide = True
gd.calculate_all()
all_features = {}

for feat, value in zip(gd.featurenames, gd.descriptor[0]):
        all_features[feat] = value
charge_all = all_features["Charge"]

print(charge_direct, charge_all)
```
